# Step 02 — Migration Planner

**Goal:** Determine the correct logical order to migrate every object, and attach structured migration flags to each one.

The planner reads the knowledge graph built in notebook 01 and answers:
- Which tables have no dependencies and can be migrated first?
- Which views depend on which tables (and must come after)?
- Which procedures/triggers must come last?
- What MySQL-specific features need special treatment per object?

**What decides the order?**

| Edge type | Meaning for ordering |
|---|---|
| `FK` (TABLE→TABLE) | Referenced table must be migrated first |
| `DEPENDS_ON` (VIEW→TABLE) | All source tables must exist before the view |
| `TOUCHES` (PROC→TABLE) | Procedure/trigger migrated after all touched tables |
